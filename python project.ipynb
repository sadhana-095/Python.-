```python
import openpyxl

# Create a new workbook object
workbook = openpyxl.Workbook()

# Add a new worksheet to the workbook
workbook.create_sheet('Sheet1')
workbook.create_sheet('Sheet2')

# Select the active worksheet
sheet = workbook.active

# Write some data to the worksheet
sheet['A1'] = 'First Name'
sheet['B1'] = 'Last Name'
sheet['C1'] = 'Email'
sheet['D1'] = 'Mobile'

# Save the workbook to a file
workbook.save('registration_data.xlsx')
```

```python
import tkinter as tk
from tkinter import messagebox

def register():
    # Get the user input from the form
    first_name = first_name_entry.get()
    last_name = last_name_entry.get()
    email = email_entry.get()
    mobile = mobile_entry.get()

    # Create a new row with the user input
    new_row = [first_name, last_name, email, mobile]

    # Append the new row to the Excel sheet
    workbook = openpyxl.load_workbook("registration_data.xlsx")
    sheet = workbook.active
    sheet.append(new_row)
    workbook.save("registration_data.xlsx")

    messagebox.showinfo("Success", "Registration successful!")

# Create the main tkinter window
root = tk.Tk()
root.title("Registration Form")
root.geometry('300x300')

# Create labels and entry fields for each input
first_name_label = tk.Label(root, text="First Name:")
first_name_label.pack()
first_name_entry = tk.Entry(root)
first_name_entry.pack()

last_name_label = tk.Label(root, text="Last Name:")
last_name_label.pack()
last_name_entry = tk.Entry(root)
last_name_entry.pack()

email_label = tk.Label(root, text="Email:")
email_label.pack()
email_entry = tk.Entry(root)
email_entry.pack()

mobile_label = tk.Label(root, text="Mobile:")
mobile_label.pack()
mobile_entry = tk.Entry(root)
mobile_entry.pack()

# Create a register button
register_button = tk.Button(root, text="Register", command=register)
register_button.pack()

root.mainloop()
```
